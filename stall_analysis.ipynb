```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# Stock information
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

```python
total_sales = sum(daily_sales)

average_sales = sum(daily_sales) / len(daily_sales)

highest_sales = max(daily_sales)

lowest_sales = min(daily_sales)

best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

best_day = days[best_index]
worst_day = days[worst_index]

best_week = (best_index // 7) + 1
worst_week = (worst_index // 7) + 1

print("Total Sales:", total_sales)
print("Average Sales:", round(average_sales, 2))
print("Highest Sales:", highest_sales)
print("Lowest Sales:", lowest_sales)

print("Best Day:", best_day, "Week", best_week)
print("Worst Day:", worst_day, "Week", worst_week)
```

```python
def week_totals(sales):
    week1 = sum(sales[:7])
    week2 = sum(sales[7:])
    
    return week1, week2


week1, week2 = week_totals(daily_sales)

growth = ((week2 - week1) / week1) * 100

print("Week 1 Sales:", week1)
print("Week 2 Sales:", week2)
print("Weekly Growth:", round(growth, 2), "%")
```

```python
def classify_day(amount, average):
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"


counts = {}

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    counts[category] = counts.get(category, 0) + 1


print("Daily Sales Classification:")
print(counts)
```

```python
weekdays = daily_sales[0:5] + daily_sales[7:12]

weekends = daily_sales[5:7] + daily_sales[12:14]

weekday_average = sum(weekdays) / len(weekdays)

weekend_average = sum(weekends) / len(weekends)

difference = ((weekend_average - weekday_average) / weekday_average) * 100

print("Weekday Average:", weekday_average)
print("Weekend Average:", weekend_average)
print("Percentage Difference:", round(difference, 2), "%")
```

```python
def products_to_reorder(stock):
    reorder_list = []

    for product, details in stock.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


reorder_products = products_to_reorder(stock)

print("Products to Reorder:", reorder_products)


print("\nGross Margin Percentages:")

for product, details in stock.items():
    cost = details["cost"]
    price = details["price"]

    margin = ((price - cost) / price) * 100

    print(product, ":", round(margin, 2), "%")
```

```python
print("=" * 45)
print("GIKOMBA STALL PERFORMANCE REPORT")
print("=" * 45)

print(f"Total Sales: KES {total_sales:,}")
print(f"Average Daily Sales: KES {average_sales:,.2f}")
print(f"Highest Sales: KES {highest_sales:,}")
print(f"Lowest Sales: KES {lowest_sales:,}")

print("-" * 45)

print(f"Week 1 Sales: KES {week1:,}")
print(f"Week 2 Sales: KES {week2:,}")
print(f"Weekly Growth: {growth:.2f}%")

print("-" * 45)

print(f"Weekday Average: KES {weekday_average:,.2f}")
print(f"Weekend Average: KES {weekend_average:,.2f}")

print("-" * 45)

print("Sales Categories:")
for category, count in counts.items():
    print(f"{category}: {count} days")

print("-" * 45)

print("Products Requiring Reorder:")
for product in reorder_products:
    print(product)

print("-" * 45)

print("Gross Margins:")
for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    print(f"{product}: {margin:.2f}%")

print("=" * 45)
print("             END OF REPORT")
print("=" * 45)
```

## Business Recommendations for Gikomba Stall

1. **Prepare for busy weekends:** Weekend sales averaged KES 4,900 compared to KES 2,685 on weekdays. Gikomba Stall should ensure enough stock is available before weekends to meet customer demand.

2. **Restock essential products:** Tomatoes and sukuma are below their reorder levels. Replenishing these products will help prevent shortages and avoid losing potential sales.

3. **Improve weekday sales:** Six out of the fourteen days recorded weak sales. The stall could try small promotions or discounts on slower days to attract more customers.

4. **Focus on profitable products:** Avocado has the highest gross profit margin at approximately 42.86%. The stall should monitor its availability and consider maintaining sufficient stock.

5. **Monitor sales growth:** Sales increased by approximately 10.66% in the second week. The stall should continue tracking its performance to identify whether this improvement can be maintained.
